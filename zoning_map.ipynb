```python
import geopandas as gpd
import json

features = []

with open("/content/Zoning_Cases.geojson", "r", encoding="utf-8", errors="ignore") as f:
    for line in f:
        if '"type": "Feature"' in line:
            try:
                obj = json.loads(line.strip().rstrip(","))
                features.append(obj)
            except:
                continue

print("Số feature đọc được:", len(features))
```

```python
gdf = gpd.GeoDataFrame.from_features(features)
gdf.head()
```

```python
import folium

gdf_sample = gdf.sample(1000)

gdf_sample = gdf_sample.set_crs(epsg=4326, allow_override=True)
gdf_sample = gdf_sample.to_crs(epsg=4326)

m = folium.Map(location=[38.9, -77.0], zoom_start=11)

folium.GeoJson(gdf_sample).add_to(m)

m
```

```python
import folium
import json

geojson_data = json.loads(gdf_sample.to_json())

m = folium.Map(location=[38.9, -77.0], zoom_start=11)

folium.GeoJson(
    geojson_data,
    tooltip=folium.GeoJsonTooltip(
        fields=["CASE_NAME", "CASE_TYPE", "ACTION_TAKEN"]
    )
).add_to(m)

m
```

```python
m.save("map_zoning.html")
```
